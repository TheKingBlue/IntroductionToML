## 2. Methods (3 points)

### 2.1 Explain Learning Vector Quantization (0.5 points)

Explain the algorithm in a general manner.

**Your answer here:** 

The idea is to create prototypes that represent each class. When identifying new data with the model, the class of the closest prototype is assigned to the presented datapoint. To train the model, the prototypes for each class must be positioned using the labeled datapoints:

- Initialize prototype vector(s) for each class
- Present each datapoint sequentially:
    - Identify the closest prototype to the datapoint (the winner)
    - Update the winner:
        - If the datapoint has the same class label as the winner, move the winner closer to the datapoint
        - If the datapoint does not have the same label, move the winner away from the datapoint

#### LVQ1 Steps

Implement the following steps:

1. **Initialize K prototypes per class, e.g., by random selection of data points**

2. **Repeat for epochs $t = 1$ to $t_{max}$:**

      *   In each epoch, present all of the dataset in a randomized order (every example is presented exactly once, the order of examples is different in every epoch). Take care that also the training labels are permuted in the very same way.

      *   perform an epoch of training using all of the P examples. At every individual step present a single example to the system, evaluate the distances from all prototypes and update the winning prototype

      *   After each epoch, compute the predicted labels of the data samples and determine the number E of misclassified training examples, i.e. the "training error".

```python
def classification_label_and_error(data, labels, prototypes, prototype_labels):
    '''
    Compute the predicted labels of the data samples and the number of misclassified samples

    Args:
        data (ndarray): Data points.
        labels (ndarray): labels of individual data points
        prototypes (ndarray): The prototypes
        prototype_labels (ndarray): the labels of the prototypes

    Returns:
        predicted_labels (list): the predicted labels of the data points
        error (int): the number of misclassified samples
    '''
    predicted_labels = []
    error = 0

    for p_i in range(len(data)):
        # Compute index of the winner and its label
        w_i = np.argmin([np.linalg.norm(w - data[p_i]) for w in prototypes])
        w_label = prototype_labels[w_i]
        # Update the predicted labels
        predicted_labels += [w_label]

        # If the predicted label is wrong, increase the error
        if w_label != labels[p_i]:
            error += 1
    
    return predicted_labels, error
```

```python
def initialize_prototypes(data, num_prototypes):
    '''
    Initializes prototypes for data evenly split between two classes.
    Where the first half is class one, and the second half class two.

    Args:
        data (ndarray): Data points
        num_prototypes (int): The number of prototypes

    Returns:
        prototypes (ndarray): The prototypes
    '''
    prototypes = data[np.random.choice(data[:len(data)//2].shape[0], num_prototypes, replace=False)]
    prototypes = np.append(prototypes, data[np.random.choice(data[len(data)//2:].shape[0], num_prototypes, replace=False)], axis=0) 
    prototype_labels = np.array([1] * num_prototypes + [2] * num_prototypes)

    return prototypes, prototype_labels


def learning_vector_quantization(data, labels, num_prototypes, learning_rate, max_epoch):
    '''
    Supervised learning vector quantization (LVQ1)

    Args:
        data (ndarray): Data points.
        labels (ndarray): labels of the data points
        num_prototypes (int): The number of prototypes per class
        learning_rate (float): The learning rate
        max_epoch (int): The maximum number of epochs

    Returns:
        prototype_trace (ndarray): the trace of the prototypes over all epochs
        predicted_labels (ndarray): the predicted labels of the data points
        num_misclassification (ndarray): the number of misclassified samples in each epoch
    '''
    prototypes, prototype_labels = initialize_prototypes(data, num_prototypes)

    prototype_trace = [prototypes.copy()]
    num_misclassification = []

    for e in range(max_epoch):
        data_indices = np.random.permutation(np.arange(0, len(data)-1))

        for p_i in data_indices:
            # Compute index of the winner
            w_i = np.argmin([np.linalg.norm(w - data[p_i]) for w in prototypes])

            # Check if the winner and p are of the same class, then update accordingly
            if prototype_labels[w_i] == labels[p_i]:
                # Pull the winner closer
                prototypes[w_i] += learning_rate * (data[p_i] - prototypes[w_i])
            if prototype_labels[w_i] != labels[p_i]:
                # Push the winner away
                prototypes[w_i] -= learning_rate * (data[p_i] - prototypes[w_i])
        
        # Update prototype_trace, predicted_labels and num_misclassification
        prototype_trace += [prototypes.copy()]
        predicted_labels, error = classification_label_and_error(data, labels, prototypes, prototype_labels)
        num_misclassification += [error]

    return prototype_trace, predicted_labels, num_misclassification
```
